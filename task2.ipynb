```python
#load class
import numpy as np
import pandas as pd 
import os
from sklearn.neighbors import KNeighborsClassifier 
from sklearn.preprocessing import StandardScaler #module for perform scaling
from sklearn.model_selection import train_test_split #module for splitting datatset
from sklearn import metrics #module for evaluating performance
```

```python
#load your data
df = pd.read_csv("databug.csv") #change the name accordingly
df.head() # prints top 5 rows from the datatset to check data is load or no
```

```python
# Check for missing values
missing_values = df.isnull().sum()
print("Missing Values:\n", missing_values)
```

```python
# remove duplicatesd
df = df.drop_duplicates()
```

```python
# prepare features
x = df.drop(['Label'],axis=1) #remove class or label
y = df['Label'] #load label
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=.2)
# Scale the data using standardization
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train) #scale training set
x_test = scaler.transform(x_test) #scale test set
z = KNeighborsClassifier(n_neighbors=4) # KNN classifier for 4 neighbours
KNN = z.fit(x_train,y_train) # start training
predict = KNN.predict(x_test) # performance in the test set
acc_knn = metrics.accuracy_score(y_test,predict)
print("Accuracy:", acc_knn) 
```

```python
# library for save and load scikit-learn models
import pickle
# file name, recommending *.pickle as a file extension
filename = "model.pickle"
# save model
pickle.dump(z, open(filename, "wb"))
```

```python
from sklearn.svm import SVC

svm_model = SVC(kernel="linear")
svm_model.fit(x_train, y_train)

# Predict and evaluate
predict = svm_model.predict(x_test)
acc_svm =  metrics.accuracy_score(y_test,predict)
print("Accuracy:", acc_svm)
```

```python
# file name, recommending *.pickle as a file extension
filename = "svm_model.pickle"
# save model
pickle.dump(svm_model, open(filename, "wb"))
```

```python
from sklearn.ensemble import RandomForestClassifier
# Initialize and train the model
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(x_train, y_train)

# Predict and evaluate
predict = rf_model.predict(x_test)
acc_rf = metrics.accuracy_score(y_test,predict)
print("Accuracy:", acc_rf)
```

```python
# file name, recommending *.pickle as a file extension
filename = "rf_model.pickle"
# save model
pickle.dump(rf_model, open(filename, "wb"))
```

```python
# format output into table
performance_table = pd.DataFrame({
    "Model": ["K-Nearest Neighbors", "Support Vector Machine", "Random Forest"],
    "Accuracy": [acc_knn, acc_svm, acc_rf],
})
performance_table["Accuracy"] = performance_table["Accuracy"].map(lambda accuracy: f"{accuracy:.2%}")
performance_table
```
